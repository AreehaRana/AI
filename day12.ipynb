```python
# Parent Class
class Shape:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("This is a", self.name)


# Child Class 1
class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


# Child Class 2
class Square(Shape):
    def __init__(self, side):
        super().__init__("Square")
        self.side = side

    def area(self):
        return self.side * self.side


# Child Class 3
class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__("Rectangle")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
#Child class 4
class Triangle(Shape):
    def __init__(self,base,height):
        super(). __init__("Triangle")
        self.base =base
        self.height = height
    def area(self):
        return 0.5* self.base*self.height
    
# input values
radius = float(input("Enter the value of radius"))
square_side = float(input("Enter the length  "))
rectangle_length = float(input("Enter the length "))
rectangle_width = float (input("Enter width of the rectangle"))




# Creating Objects
c = Circle(radius)
s = Square(square_side)
r = Rectangle(rectangle_length,rectangle_width)
t = Triangle(10,5)

# Calling Methods
c.display()
print("Area:", c.area())

s.display()
print("Area:", s.area())

r.display()
print("Area:", r.area())

t.display()
print("Area:" , t.area())
```

```python
# Parent Class
class Employee:
    def __init__(self, name, employee_id, salary):
        self.__name = name                # Private attribute
        self.__employee_id = employee_id  # Private attribute
        self.__salary = salary            # Private attribute
        self._department = "General"      # Protected attribute

    # Getter Methods
    def get_name(self):
        return self.__name

    def get_employee_id(self):
        return self.__employee_id

    def get_salary(self):
        return self.__salary

    # Setter Method
    def set_salary(self, new_salary):
        if new_salary > 0:
            self.__salary = new_salary
        else:
            print("Salary must be positive.")

    # Bonus Calculation
    def calculate_bonus(self):
        return self.__salary * 0.05  # 5% bonus

    # Display Info
    def display_info(self):
        print("Name:", self.__name)
        print("Employee ID:", self.__employee_id)
        print("Salary:", self.__salary)
        print("Department:", self._department)


# Child Class
class Manager(Employee):
    def __init__(self, name, employee_id, salary, team_size):
        super().__init__(name, employee_id, salary)
        self.__team_size = team_size      # Private attribute
        self._department = "Management"   # Modify protected attribute

    # Getter for team size
    def get_team_size(self):
        return self.__team_size

    # Method Overriding
    def calculate_bonus(self):
        # Managers get 10% bonus instead of 5%
        return self.get_salary() * 0.10

    # Overriding display method
    def display_info(self):
        super().display_info()
        print("Team Size:", self.__team_size)


# Object Creation
e1 = Employee("Ali", "E123", 50000)
m1 = Manager("Bob", "M456", 90000, 5)

print(" Employee Info ")
e1.display_info()
print("Bonus:", e1.calculate_bonus())

print("Manager Info")
m1.display_info()
print("Bonus:", m1.calculate_bonus())

        
```

```python
class BonusCalculator:
    def __init__(self, salary):
        self.__salary = salary
        self.__bonus_percent = 5   
    def set_salary(self, salary):
        self.__salary = salary
    def get_salary(self):
        return self.__salary
    def get_bonus_percent(self):
        return self.__bonus_percent
    def calculate_bonus(self):
        if self.__salary > 5000:
            return (self.__salary * self.__bonus_percent) / 100
        else:
            return 0
    def total_salary(self):
        return self.__salary + self.calculate_bonus()
    def display(self):
        print("Salary:", self.get_salary())
        if self.get_salary() > 5000:
            print("Bonus %:", self.get_bonus_percent())
            print("Bonus Amount:", self.calculate_bonus())
        else:
            print("No Bonus (Salary must be above 5000)")
        print("Total Salary:", self.total_salary())
salary = float(input("Enter Salary: "))
emp1 = BonusCalculator(salary)
emp1.display()
```
