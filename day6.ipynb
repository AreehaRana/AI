```python
#1.ATM Withdrawal Check
balance = 5000
amount = int(input("Enter withdrawal amount: "))

if amount <= balance:
    print("Transaction Successful ")
else:
    print("Insufficient Balance ")
```

```python
#2.Electricity Bill Calculator
units = int(input("Enter units used: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = units * 8
else:
    bill = units * 10

print("Total Bill:", bill)
```

```python
#3.Check Number Type (Single/Double/Triple Digit)
num = int(input("Enter number: "))

if 0 <= num <= 9:
    print("Single Digit")
elif 10 <= num <= 99:
    print("Double Digit")
elif 100 <= num <= 999:
    print("Triple Digit")
else:
    print("More than 3 digits")
```

```python
#4.BMI Category Checker
bmi = float(input("Enter BMI: "))

if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

```python
#5. Password strength checker
password = input("Enter password: ")

if len(password) >= 8:
    print("Strong Password ")
elif len(password) >= 5:
    print("Medium Password")
else:
    print("Weak Password")
```

```python
#6. check day type (weekend or weekday)
day = input("Enter day: ").lower()

if day == "saturday" or day == "sunday":
    print("Weekend 🎉")
elif day in ["monday", "tuesday", "wednesday", "thursday", "friday"]:
    print("Weekday ")
else:
    print("Invalid Day")
```

```python
#7.Shopping Discount Calculator
amount = float(input("Enter shopping amount: "))

if amount >= 5000:
    discount = 0.20
elif amount >= 2000:
    discount = 0.10
else:
    discount = 0

final_amount = amount - (amount * discount)
print("Final Amount:", final_amount)
```

```python
#8.check Alphabet case
ch = input("Enter a character: ")

if ch.isupper():
    print("Uppercase Letter")
elif ch.islower():
    print("Lowercase Letter")
else:
    print("Not a letter")
```

## Tasks: Break, Continue, Pass

```python
## 1. Print numbers until 10 but stop at 7
for i in range(1, 11):
    if i == 7:
        break
    print(i)


```

```python
#2. Print even numbers until 20 but skip 10
for i in range(1, 21):
    if i == 10:
        continue
    if i % 2 == 0:
        print(i)
        
```

```python
#3. Loop through numbers 1–5 but use pass for number 3  
for i in range(1, 6):
    if i == 3:
        pass
    else:
        print(i)
        
```

```python
#4. Keep asking user for numbers and stop when 0 is entered
while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    print("You entered: ", num)                                                                    
```

```python
#5. Print only odd numbers from 1 to 20
for i in range(1, 21):
    if i % 2 != 0:
        print(i)        
        
```

```python
#6. Add numbers until total becomes greater than 100

total = 0
while total <= 100:                                                 
    num = int(input("Enter a number: "))
    total += num
print("Total exceeded 100! Final Total:", total)


        
```

```python
#7. Print only consonants from a string
vowels = "aeiouAEIOU"
string = input("Enter a string: ")
for char in string:
    if char not in vowels and char.isalpha():
        print(char)


    
```

```python
#8. Infinite loop that stops when user types "exit"
while True:
    user_input = input("Enter text (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("You entered:", user_input)
```

```python
#9. Print numbers from 1 to 30 but skip multiples of 5
for i in range(1, 31):
    if i % 5 == 0:
        continue
    print(i)
```

```python
#10. Loop through a list of names and use pass if name starts with "A"
names = ["Alice", "Bob", "Charlie", "David", "Eve"]
for name in names:
    if name.startswith("A"):
        pass
    else:
        print(name)
        
```

## Combination of If Elif Else Break Continue

```python
#1. Number game
while True:
    num = int(input("Enter a number (0 to exit): "))
    
    if num == 0:
        print("Game Over! ")
        break
    elif num < 0:
        print("Negative skipped ")
        continue
    elif num % 5 == 0:
        pass  # do nothing
    elif num % 2 == 0:
        print(f"{num} is Even ")
    else:
        print(f"{num} is Odd ")
```

```python
#Shopping discount 
while True:
    amount = float(input("Enter shopping amount (0 to exit): "))
    
    if amount == 0:
        break
    elif amount < 0:
        continue
    elif amount >= 5000:
        print("20% Discount ")
    elif amount >= 2000:
        print("10% Discount ")
    else:
        pass  # no discount
```
