## Tasks: Break, Continue, Pass

```python
# 1. Print numbers until 5 using break
for i in range(1, 6):
    if i == 5:
        break
    print(i)
```

```python
# 2. Skip number 3 using continue
for i in range(1,6):
    if i ==3:
        continue
    print(i)
```

```python
# 3. Loop with pass as placeholder
for i in range (5):
    pass 
```

```python
# 4. Stop loop when a negative number appears
numbers = [5, 8, 3, -2, 7, 9]
for num in numbers:
    if num < 0:
        print("Negative number found. Loop stopped.")
        break
    print(num)
```

```python
# 6. Break when sum exceeds 50
total = 0

for i in range(1, 21):
    total += i
    if total > 50:
        print("Sum exceeded 50")
        break
    print("Current sum:", total)
```

```python
# 7. Continue for vowels only in a string
text = input("Enter a string:")
for ch in text:
    if ch.lower() not in "aeiou":
        continue
        print (ch)
```

```python
# 8. Infinite loop that exits with break
count = 1

while True:   
    print("Count:", count)
    
    if count == 5:
        break   
    
    count += 1
```

```python
# 9. Skip multiples of 4
for i in range(1, 21):
    if i % 4 == 0:
        continue   
    print(i)
```

```python
# 10. Use pass inside an if condition
for i in range(1, 6):
    if i == 3:
        pass   # Do nothing when i is 3
    else:
        print(i)
```
