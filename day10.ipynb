```python
#2.Create a class Rectangle Attributes: length, width Method: area() → returns areaMethod: perimeter() → returns perimeter
class Rectangle:
    def __init__( self,length, width):
        self.length= length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
r1 = Rectangle(10, 5)

print("Area:", r1.area())
print("Perimeter:", r1.perimeter())
```

```python
#4.Create a class Student Attribute: marks (dictionary of subjects) Method: average() → returns average marks
class Student:
    def __init__(self, marks):
        self.marks = marks   

    def average(self):
        total = sum(self.marks.values())
        count = len(self.marks)
        return total / count

marks_data = {
    "Math": 90,
    "English": 80,
    "Science": 85
}

s1 = Student(marks_data)

print("Average Marks:", s1.average())
        
```

```python
#6.Create a class BankAccount
#Attribute: balance
#Method: deposit(amount)Method: withdraw(amount)
#Do not allow withdrawal if balance is insufficient
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount #balance = balance + amount 
        print("Amount Deposited:", amount)
        print("New Balance:", self.balance)

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient Balance!")
        else:
            self.balance -= amount
            print("Amount Withdrawn:", amount)
            print("New Balance:", self.balance)

account1 = BankAccount(1000)
account1.deposit(500)
account1.withdraw(300)
account1.withdraw(2000)
```

```python
#8 Create a class Laptop Attribute: brand, price Method: apply_discount(percent)Reduce price by given percent
class Laptop:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def apply_discount(self, percent):
        discount = (percent / 100) * self.price
        self.price -= discount
        return self.price


# Creating object
l1 = Laptop("Dell", 50000)
print("Price before discount:", l1.price)
l1.apply_discount(10)
print("Price after discount:", l1.price)
          
```

```python
#9 Create a class Password Attribute: password Method: check_strength()
#If length < 6 → Weak
#If length 6–10 → Medium
#If length > 10 → Strong
class Password:
    def __init__(self, password):
        self.password = password

    def check_strength(self):
        length = len(self.password)

        if length < 6:
            return "Weak"
        elif 6 <= length <= 10:
            return "Medium"
        else:
            return "Strong"


p1 = Password("abc123")
print("Password Strength:", p1.check_strength())

p2 = Password("mysecurepassword123")
print("Password Strength:", p2.check_strength())

```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
        self.transactions = []

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount!")
            return
        self.balance += amount
        self.transactions.append(f"Deposited: {amount}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount!")
            return
        if amount > self.balance:
            print("Insufficient Balance!")
            return
        self.balance -= amount
        self.transactions.append(f"Withdrawn: {amount}")

    def transfer(self, other_account, amount):
        if amount <= 0:
            print("Invalid transfer amount!")
            return
        if amount > self.balance:
            print("Insufficient Balance for transfer!")
            return
        
        self.balance -= amount
        other_account.balance += amount
        
        self.transactions.append(f"Transferred {amount} to {other_account.owner}")
        other_account.transactions.append(f"Received {amount} from {self.owner}")

    def show_balance(self):
        print(f"{self.owner}'s Balance: {self.balance}")

    def show_transactions(self):
        print(f"\nTransaction History of {self.owner}:")
        for t in self.transactions:
            print(t)
acc1 = BankAccount("Areeha", 5000)
acc2 = BankAccount("Ali", 3000)

acc1.deposit(2000)
acc1.withdraw(1000)
acc1.transfer(acc2, 1500)

acc1.show_balance()
acc2.show_balance()

acc1.show_transactions()
acc2.show_transactions()
```

```python
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def subtract(self):
        return self.num1 - self.num2

    def multiply(self):
        return self.num1 * self.num2

    def divide(self):
        if self.num2 == 0:
            return "Error! Division by zero is not allowed."
        return self.num1 / self.num2


# Creating object
calc = Calculator(10, 5)

print("Addition:", calc.add())
print("Subtraction:", calc.subtract())
print("Multiplication:", calc.multiply())
print("Division:", calc.divide())

```

```python
class Movies:
    def __init__(self):
        self.movie_list = []

        n = int(input("How many movies do you want to enter? "))

        for i in range(n):
            movie = input("Enter movie name: ")
            self.movie_list.append(movie)

    def show_movies(self):
        print("Movie List:")
        for movie in self.movie_list:
            print(movie)


# Creating object
m1 = Movies()

# Showing movies
m1.show_movies()
```
