```python
# 7. Shopping Discount Calculator

def calculate_discount():
    amount = float(input("Enter shopping amount: "))

    if amount >= 5000:
        discount = 0.20
    elif amount >= 2000:
        discount = 0.10
    else:
        discount = 0

    final_amount = amount - (amount * discount)
    print("Final Amount:", final_amount)

calculate_discount()
```
