## Encapsulation

```python
class Student:
    def __init__(self):
        self.__name = "ali"     # private variable
        self.age = 20

    def display(self):
        print("Name:", self.__name)
        print("Age:", self.age)

    def get_name(self):
        return self.__name

    def set_name(self, name):   # parameter added
        self.__name = name

s1 = Student()
print("First time:")
s1.display()
print("Try to change directly:")
s1.__name = "gurriya"   
print("Get name:", s1.get_name())
print("Change using setter:")
s1.set_name("gurriya")
s1.display()
```

```python
## bonus calculate
class BonusCalculator:
    
    def __init__(self, salary):
        self.salary = salary
        self.bonus_percent = 5   # Fixed 5% bonus
    
    def calculate_bonus(self):
        bonus = (self.salary * self.bonus_percent) / 100
        return bonus
    
    def total_salary(self):
        return self.salary + self.calculate_bonus()
    
    def display(self):
        print("Salary:", self.salary)
        print("Fixed Bonus %:", self.bonus_percent)
        print("Bonus Amount:", self.calculate_bonus())
        print("Total Salary:", self.total_salary())

salary = float(input("Enter Salary: "))

emp1 = BonusCalculator(salary)

emp1.display()
        
```

```python
class BonusCalculator:
    def __init__(self, salary):
        self.__salary = salary
        self.__bonus_percent = 5   
    def set_salary(self, salary):
        self.__salary = salary
    def get_salary(self):
        return self.__salary
    def get_bonus_percent(self):
        return self.__bonus_percent
    def calculate_bonus(self):
        if self.__salary > 5000:
            return (self.__salary * self.__bonus_percent) / 100
        else:
            return 0
    def total_salary(self):
        return self.__salary + self.calculate_bonus()
    def display(self):
        print("Salary:", self.get_salary())
        if self.get_salary() > 5000:
            print("Bonus %:", self.get_bonus_percent())
            print("Bonus Amount:", self.calculate_bonus())
        else:
            print("No Bonus (Salary must be above 5000)")
        print("Total Salary:", self.total_salary())
salary = float(input("Enter Salary: "))
emp1 = BonusCalculator(salary)
emp1.display()
```

```python
class Student:
    
    def __init__(self, age):
        self.__age = age   # private variable

    def set__age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Invalid Age")

    def get_age(self):
        return self.__age
age = int(input("Enter your age: "))
s1 = Student(age)
s1.set__age(-50) 
print("Age:", s1.get_age())
s2 = Student(25)
print("Second Student Age:", s2.get_age())

```

```python
class Car:
    def __init__(self, speed):
        self.__speed = speed
    def set_speed(self, speed):
        if 0 <= speed <= 200:
            self.__speed = speed
            print("Your speed is", self.__speed)
        else:
            print("Over speed! Speed must be between 0 and 200.")

    def get_speed(self):
        return self.__speed
speed = int(input("Enter your speed: "))
s1 = Car(speed)
new_speed = int(input("Enter new speed: "))
s1.set_speed(new_speed)
print("Current speed:", s1.get_speed())
```
