```python
#Create a base class Animal with method sound().Create subclasses:Dog Cat CowEach overrides sound() and prints a different sound.
class Animal:
    def sound(self):
        print("the sound of the animaal")
class Dog(Animal):
    def sound(self):
        print("the dog says waooo")
class Cat(Animal):
    def sound(self):
        print("cat say meoww")
class Cow(Animal):
    def sound (Self):
        print("the cow says mooooo")
A1 = Animal()
D1 = Dog()
C1 = Cat()
C2 = Cow()

A1.sound()
D1.sound()
C1.sound()
C2.sound()
```

```python
# Shape Area Calculation.Create base class Shape with method area().subclasses:• Rectangle (length × width)• Circle (πr²)• Triangle (½ × base × height)Override area() in each class.
class Shape:
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height
shapes = [
    Rectangle(2, 2),
    Circle(3),
    Triangle(2, 5)
]
for shape in shapes:
    print("Area:", shape.area())

    
        
```

```python
#Base class Payment with method pay(amount).Subclasses:• CreditCardPayment• JazzCashPayment• BankTransferPaymentEach should print a different transaction message
class Payment:
    def pay(self, amount):
        self.amount = amount
        print("Processing payment of", self.amount)
class CreditCardPayment(Payment):
    def pay(self, amount):
        print("This is credit card payment of", amount)
class JazzCashPayment(Payment):
    def pay(self, amount):
        print("This is JazzCash payment of", amount)
class BankTransferPayment(Payment):
    def pay(self, amount):
        print("This is bank transfer payment of", amount)
P1 = [CreditCardPayment(), JazzCashPayment(), BankTransferPayment()]
for payment in P1:
    payment.pay(100)
```

```python
# Base class Employee with method calculate_salary().
# Subclasses:
# • FullTimeEmployee (fixed salary)
# • PartTimeEmployee (hourly_rate × hours)
# • Intern (stipend based)
# Call calculate_salary() using polymorphism.
class Employee:
    def calculate_salary(self):
        pass
class FullTimeEmployee(Employee):
    def __init__(self, fixed_salary):
        self.fixed_salary = fixed_salary

    def calculate_salary(self):
        return self.fixed_salary
class PartTimeEmployee(Employee):
    def __init__(self, hourly_rate, hours):
        self.hourly_rate = hourly_rate
        self.hours = hours

    def calculate_salary(self):
        return self.hourly_rate * self.hours
class Intern(Employee):
    def __init__(self, stipend):
        self.stipend = stipend

    def calculate_salary(self):
        return self.stipend
employees = [
    FullTimeEmployee(10000),
    PartTimeEmployee(10, 10),
    Intern(2000)
]
for employee in employees:
    print("Salary:", employee.calculate_salary())
    
        


```

```python
#Base class Customer with method get_discount(amount).
#Subclasses:
#• RegularCustomer (5%)
#• PremiumCustomer (10%)
#• VIPCustomer (20%)
#Return final payable amount
class Customer:
    def get_discount(self, amount):
        pass
class RegularCustomer(Customer):
    def get_discount(self, amount):
        return amount * 0.05
class PremiumCustomer(Customer):
    def get_discount(self, amount):
        return amount * 0.10
class VIPCustomer(Customer):
    def get_discount(self, amount):
        return amount * 0.20
customers = [
    RegularCustomer(),
    PremiumCustomer(),
    VIPCustomer()
]
amount = 1000
for customer in customers:
    discount = customer.get_discount(amount)
    print("Discount:", discount)
```

```python
class Account:
    def withdraw(self, amount):
        pass


class SavingsAccount(Account):
    def withdraw(self, amount):
        fee = amount * 0.02
        total = amount + fee
        print("Savings Account: Withdraw", amount, "+ 2% fee =", total)


class CurrentAccount(Account):
    def withdraw(self, amount):
        print("Current Account: Withdraw", amount, "(No fee)")


class BusinessAccount(Account):
    def withdraw(self, amount):
        fee = amount * 0.05
        total = amount + fee
        print("Business Account: Withdraw", amount, "+ 5% fee =", total)


accounts = [
    SavingsAccount(),
    CurrentAccount(),
    BusinessAccount()
]

amount = 1000

for account in accounts:
    account.withdraw(amount)
```
