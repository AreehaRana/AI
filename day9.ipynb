```python
#2. Init Function
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Ali", 20)
s2 = Student("Sara", 22)

print(s1.name)
print(s2.age)
```

```python
class Animal:
    def __init__(self):
        self.name = input("Enter animal name: ").lower()

    def sound(self):
        if self.name == "dog":
            print("Dog says: Bark")
        elif self.name == "cat":
            print("Cat says: Meow")
        elif self.name == "cow":
            print("Cow says: Moo")
        elif self.name == "lion":
            print("Lion says: Roar")
        else:
            print("Sorry! I don't know this animal's sound.")
a1 = Animal()
a1.sound()
```

```python
class Animal:
    def __init__(self):
        self.animals_list = ["dog", "cat", "cow", "lion"]
        self.name = input("Enter animal name: ").lower()

    def sound(self):
        if self.name in self.animals_list:
            print("Animal exists in list ")
            
            if self.name == "dog":
                print("Dog says: Bark")
            elif self.name == "cat":
                print("Cat says: Meow")
            elif self.name == "cow":
                print("Cow says: Moo")
            elif self.name == "lion":
                print("Lion says: Roar")
        else:
            print("Animal doesn't exist in list ")

a1 = Animal()
a1.sound()
```
