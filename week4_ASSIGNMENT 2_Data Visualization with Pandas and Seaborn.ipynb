## ASSIGNMENT 2: Data Visualization with Pandas and Seaborn

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'product_a': [200, 220, 250, 270, 300, 320],
    'product_b': [150, 180, 200, 210, 240, 260],
    'product_c': [100, 120, 140, 160, 180, 200],
    'region': ['North', 'South', 'East', 'West', 'North', 'South']
}

df = pd.DataFrame(data)

# Line Plot
df.set_index('month')[['product_a', 'product_b', 'product_c']].plot()
plt.title("Product Sales Over Months")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

# Bar Plot
df.plot(x='month', y='product_a', kind='bar')
plt.title("Product A Sales by Month")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

# Pie Chart
region_sales = df.groupby('region')[['product_a','product_b','product_c']].sum().sum(axis=1)

region_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Total Sales by Region")
plt.ylabel("")
plt.show()

#Seaborn Distribution Plots

# Distplot
sns.histplot(df['product_a'], kde=True)
plt.title("Distribution of Product A")
plt.show()

# Jointplot
sns.jointplot(x='product_a', y='product_b', data=df, kind='scatter')
plt.suptitle("Product A vs Product B", y=1.02)
plt.show()

# Seaborn Categorical Plots
# Boxplot
sns.boxplot(x='region', y='product_a', data=df)
plt.title("Product A Distribution by Region")
plt.show()

# Barplot (average sales by region)
sns.barplot(x='region', y='product_a', data=df)
plt.title("Average Product A Sales by Region")
plt.show()

#Matrix Plot (Heatmap)

corr = df[['product_a','product_b','product_c']].corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()
```
