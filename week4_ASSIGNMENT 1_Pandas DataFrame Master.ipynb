##ASSIGNMENT 1: Pandas DataFrame Master

```python
import pandas as pd
import numpy as np

# Step 1: Create Employees DataFrame
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4],
    'name': ['Ali', 'Bilal', 'Ahmed', 'Sara'],
    'dept_id': [101, 102, 101, 103]
})

# Step 2: Create Departments DataFrame
departments = pd.DataFrame({
    'dept_id': [101, 102, 104],
    'dept_name': ['HR', 'IT', 'Finance']
})

# Step 3: Create Salaries DataFrame
salaries = pd.DataFrame({
    'emp_id': [1, 2, 3, 4],
    'salary': [50000, 70000, 90000, 60000]
})

# Step 4: Merging (Inner, Left, Outer)

inner_join = pd.merge(employees, departments, on='dept_id', how='inner')
print("Inner Join:\n", inner_join)

left_join = pd.merge(employees, departments, on='dept_id', how='left')
print("\nLeft Join:\n", left_join)

outer_join = pd.merge(employees, departments, on='dept_id', how='outer')
print("\nOuter Join:\n", outer_join)

# Step 5: Concatenate new employees
new_employees = pd.DataFrame({
    'emp_id': [5, 6],
    'name': ['Zara', 'Usman'],
    'dept_id': [102, 104]
})

employees = pd.concat([employees, new_employees], ignore_index=True)
print("\nAfter Concatenation:\n", employees)

# Step 6: Merge employees with salaries
emp_salary = pd.merge(employees, salaries, on='emp_id', how='left')

# Step 7: Merge with departments
emp_salary = pd.merge(emp_salary, departments, on='dept_id', how='left')

print("\nEmployee + Salary + Department:\n", emp_salary)

# Step 8: GroupBy (Average Salary by Department)
avg_salary = emp_salary.groupby('dept_name')['salary'].mean()
print("\nAverage Salary by Department:\n", avg_salary)

# Step 9: Create Salary Categories using pd.cut()
bins = [0, 60000, 80000, 100000]
labels = ['Low', 'Medium', 'High']

emp_salary['salary_category'] = pd.cut(emp_salary['salary'], bins=bins, labels=labels)

print("\nWith Salary Category:\n", emp_salary)

# Step 10: Save to CSV
emp_salary.to_csv('employee_summary.csv', index=False)

print("\nCSV file 'employee_summary.csv' saved successfully!")
```
