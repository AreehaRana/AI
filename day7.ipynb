```python
# 3.Make a function that takes two numbers and prints their sum.Call the function.

def add_numbers(num1, num2):
    print("The sum of two numbers is", num1 + num2)

# Take input from user
num1 = int(input("Enter 1st number: "))
num2 = int(input("Enter 2nd number: "))

# Function calling
add_numbers(num1, num2)
```

```python
# 4.Make a function that checks if a number is even or odd.Call it with a number.

def check_even_odd(num):
    if num % 2 == 0:
        print(num, "is Even")
    else:
        print(num, "is Odd")

# Call the function with a number
check_even_odd(7)
check_even_odd(12)
```

```python
# 9. Make a function that finds the largest of two numbers.

def largest(num1, num2):
    if num1 > num2:
        print(num1, "is the largest")
    elif num2 > num1:
        print(num2, "is the largest")
    else:
        print("Both numbers are equal")

# Function calling
largest(10, 20)
largest(15, 5)
largest(7, 7)
```
