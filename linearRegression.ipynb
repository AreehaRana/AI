```python
from ast import Expression

import pandas as pd
import numpy as np
from sklearn. linear_model import LinearRegression
```

```python
X = [[2],[3],[4],[5]]
y = [10,25,35,50]
model = LinearRegression()
model.fit (X,y)
hours = float(input("Enter Study hours"))
predicted_marks = model.predict([[hours]])
print(f"Based on your hours", hours , "you may score around",predicted_marks)
```

```python
X = [[2],[3],[4],[5]] # Hour study input
y = [0,0,1,1] # result 0 F 1 P
model = LogisticRegression()
model.fit(X,y)
hours = float(input("Enter Study hours"))
result= model.predict([[hours]])[0]
if result == 1 :
    print(f"based on hours",hours,"you are likly to pass")
else:
    print(f"based on hours ", hours,"you are likly to fail")
```
